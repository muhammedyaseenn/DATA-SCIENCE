```python
class movie:
  def __init__(self,name,moviename,seatnum):
    self.name=name
    self.moviename= moviename
    self.seatnum= seatnum
  def book_ticket(self):
    print("\n",self.name," ticket booked for",self.moviename)
  def show_ticket(self):
    print("\nname        :",self.name)
    print("movie name  :",self.moviename)
    print("seat number :",self.seatnum)
p1=movie("yaseen","bulti","D3")
p2=movie("adhi","sarwam maya","D2")
p1.book_ticket()
p1.show_ticket()
p2.book_ticket()
p2.show_ticket()

```
