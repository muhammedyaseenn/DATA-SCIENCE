```python
 #here we creating a parent class

class Animal:
  def eat(self):
    print("this animal eats food")

#after create the child class
class Dog(Animal):
  def bark(self):
    print("Dog barks")

#After parent,child class object creation
d = Dog()
d.eat() #inherited from animal
d.bark() #Dog's own method
```

```python
class user:
  def login(self):
    print("user have logged in")


class student(user):
  def watch_video(self):
    print("student have watched videos")


class Teacher(user):
  def upload_video(self):
    print("teacher have uploaded video")


s = student()
t = Teacher()
s.login()
s.watch_video()
t.login()
t.upload_video()

```

```python
class father:
  def skill(self):
    #method of father class
    print("Father's skill: Gardening")

class mother:
  def skill2(self):
    print("Mother's skill: Cooking")

class child(father,mother):
  #child inherits from father and mother
  pass

#creating object

c = child()
c.skill()
c.skill2()
```

```python
#multi level

#grandparent class
class grandparent:
  def display(self):
    print("grandparent class")

#parent class inherits from grandparent class
class parent(grandparent):
  def show(self):
    print("parent class")


#child class inherits from parent class
class child(parent):
  def see(self):
    print("child class")

# create object of child
obj = child()
obj.display() #method from grandparent
obj.show()    #
obj.see()
```

```python
#base class
class animal:
  def eat(self):
    print("animal eats")

#derived class (single inheritence)
class mammal(animal):
  def walk(self):
    print("mammal walks")


class pet:
  def play(self):
    print()


```
