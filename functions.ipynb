```python
#keyword arguments
def book_details(title,author):
  print("title  :  ",title)
  print("author :  ",author)

book_details(title="python basics",author="john smith")
```

```python
def student_details(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}:{value}")

student_details(name="rahul",age=21,city="kochi")
```
