```python
my_dict={"name":"Alice","age":20,"city":"thrissur"}
print(my_dict["name"])
print(my_dict["city"])
```

```python
#modify
my_dict={"name":"Alice","age":20,"city":"thrissur"}
my_dict["age"]=21
my_dict["city"]="London"
my_dict
```

```python
#delete
my_dict={"name":"Alice","age":20,"city":"thrissur"}
del my_dict["age"]
my_dict
```

```python
#pop
my_dict={"name":"Alice","city":"thrissur"}
removed_city=my_dict.pop("city")
my_dict
```

```python
#clear()
my_dict={"name":"Alice","age":20,"city":"thrissur"}
my_dict.clear()
my_dict
```

```python
#len()
my_dict={"name":"Alice","age":20,"city":"thrissur"}
len(my_dict)
```

```python
#keys()
my_dict={"name":"Alice","age":20,"city":"thrissur"}
my_dict.keys()
```

```python
#values()
my_dict={"name":"Alice","age":20,"city":"thrissur"}
my_dict.values()
```

```python
#items()
my_dict={"name":"Alice","age":20,"city":"thrissur"}
my_dict.items()
```
