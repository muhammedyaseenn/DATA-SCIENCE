```python
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

print("\nfirst 3 rows")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))

print("\nDataFrame shape:")
print(df.shape)

print("\nDataFrame columns:")
print(df.columns)

print("\nMarks column:")
print(df["Marks"])

print("\nfirst 3 rows:")
print(df.iloc[0:3])

print("\nrows 1 to 3,Name and Marks:")
print(df.loc[1:3,["Name","Marks"]])

print("\nstudents with marks greater than 75:")
print(df[df["Marks"]>75])

print("\nstudents from kochi:")
print(df[df["city"]=="Kochi"])

print("\nstudents with marks > 70 and attendance > 80:")
print(df[
         (df["Marks"]>70) &
         (df["attendance"]>80)
]
)
```

```python
#mathematical operations
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

print("\nMarks + 5:")
print(df["Marks"]+5)

print("\nMarks * 2:")
print(df["Marks"]*2)

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nMaximum Mark:")
print(df["Marks"].max())

print("\nMinimum Mark:")
print(df["Marks"].min())
```

```python
#sorting
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

print("\nSorting by Marks:")
print(df.sort_values("Marks"))

print("\nSorted by Marks - Highest first:")
print(df.sort_values("Marks",ascending=False))
```

```python
#renaming columns
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

df_renamed=df.rename(
    columns={
        "Marks":"Score",
        "Attendance":"Attendance_percentage"
    }
)

print("\nRenamed DataFrame:")
print(df_renamed)
```

```python
#checking missing values
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

print("\nMissing values:")
print(df.isna())

print("\nNumber of missing values")
print(df.isna().sum())
```

```python
#forward fill and backward fill
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","charlie","David","Emma"],
    "Age":[20,22,19,22,20],
    "Marks":[85,72,np.nan,90,65],
    "attendance":[95,88,75,np.nan,88],
    "city":["Kochi","Delhi","Mumbai","Kochi","chennai"]
     }
df=pd.DataFrame(data)
print("orginal DataFrame:")
print(df)

df_forward=df.copy()

df_forward=df_forward.ffill()

print("\nforward fill:")
print(df_forward)
```
