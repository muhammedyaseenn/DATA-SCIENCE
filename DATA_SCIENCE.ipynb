***NUMPY [NUMERICAL PYTHON]***

 It is a python library used for working with array

 used for mathematical operation

 import - import numpy as np

```python
my_list1=[1,2,3,4]
my_list2=[2,4,6,8]
result=my_list1+my_list2
print(result)
```

```python
import numpy as np
array1=np.array([1,2,3,4])
array2=np.array([2,4,6,8])
result=array1+array2
print(result)
```

```python
import numpy as np
array1=np.array([1,2,3,4])
print(array1[1])
```

```python
#1D array
import numpy as np
array1=np.array([1,2,3,4])
print(array1[1:3])
print(array1.ndim)
```

```python
#2D array
import numpy as np
array1=np.array([[1,2,3,4],
                [3,5,7,9]])
print(array1.ndim)
```

```python
#3D array

import numpy as np
array1=np.array([[[1,2,3,4],
                [3,5,7,9],
                 [2,1,3,4]]])
print(array1.ndim)
```

```python
import numpy as np
array1=np.array([[[1,2,3,4],
                [10,20,30,40]],

                [[100,200,300,400],
                [1000,2000,3000,4000]]])
print(array1.ndim)
print(array1[1,1,3])
```

```python
#conversion of 1D array to 2D array
import numpy as np
array1=np.array([1,2,3,4,5,6,7,8])
array2=array1.reshape(2,4)
print(array1)
print(array2)
```

```python
#zeros()
a=np.zeros(5)
print(a)
```

```python
#ones()
a=np.ones(5)
print(a)
```

```python
# arange()
a=np.arange(1,10)
print(a)
```

```python
#linspace
a=np.linspace(0,10)
print(a)
```

```python
#random
a=np.random.rand(7)
print(a)
```

```python
import numpy as np
array1=np.array([[[1,2,3,4],
                  [3,5,7,9],
                  [2,1,3,4]]])

print(array1[1,1,3])
```

```python
array1=np.array([1,2,3,4,5,6])
array2=array1.reshape(2,3)
print(array2)
```

# ***series***

series is an one dimensional labelled array


```python
import pandas as pd
marks=pd.Series([1,2,3,4,5,6])
print(marks)
```

```python
import pandas as pd
a=pd.Series([2,4,5,3],index=["maths","physics","biology","hindi"])
print(a)
```

```python
import pandas as pd
a=pd.Series(["BCA","BCOM","BSC"],index=["yaseen","anshaf","shihan"])

print(a)
```

```python
import pandas as pd
a=pd.Series(["BCA","BCOM","BSC"],index=["yaseen","anshaf","shihan"])
print(a["yaseen"])
```
