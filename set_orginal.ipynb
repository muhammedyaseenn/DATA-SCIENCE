```python
#add
s=set([1,2,3,4])
s.add(5)
print(s)
```

```python
#remove
s=set([1,2,3,4,5])
s.remove(2)
print(s)
```

```python
#discard
s=set([1,3,4,5])
s.discard(10)
print(s)
```

```python
#pop
s=set([1,3,4,5])
s.pop()
print(s)
```

```python
#clear
s=set([3,4,5])
s.clear()
print(s)
```
