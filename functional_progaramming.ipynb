```python
letters=['a','b','d','e','i','j','o']

# the function that return true if letters is vowel
def filter_vowels(letter):
  vowels=['a','e','i','o','u']
  if letter in vowels:
    return True
  else:
    return False
#select only vowel elements
filter_vowels=tuple(filter(filter_vowels,letters))

print(filter_vowels)
```

```python
#filter
numbers=[1,2,3,6,8,9,10]

def filter_numbers(x):
  if x%3==0:
    return True
  else:
    return False
value=set(filter(filter_numbers,numbers))

print(value)
```

```python
#map
numbers=[1,2,3,4]

def square(x):
  return x*x

result=list(map(square,numbers))
print(result)
```

```python
numbers=[1,2,3,4,5]

def square(x):
  return x*x,x*x*x

ans=list(map(square,numbers))
print(ans)
```

```python
#map with lambda
numbers=[1,2,3,4]
result=map(lambda x:x*x,numbers)
print(result)

print(set(result))
```

```python
#add multiple lists using map()and lambda
mark1=[10,20,30]
mark2=[34,25,12]

result=map(lambda n1,n2:n1+n2,mark1,mark2)
print(list(result))
```

```python
from functools import reduce
def add(x,y):
  return x + y

a=[1,2,3,4,5]
res = reduce(add,a)
print(res)
```

```python
from functools import reduce
li=["python","is","greatfull"]

def join_word(x,y):
  return x +" "+ y
res=reduce(join_word,li)
print(res)
```

```python
#RECURSION
def factorial(n):
  if n == 1:
    return 1
  return n * factorial(n-1)

factorial(5)
```

```python
def sum_n(n):
  if n == 0:
    return 0
  return n + sum_n(n-1)

sum_n(5)
```

```python
def number_n(n):
  if n == 1:
    return 1
  return n + number_n(n-1)


number_n(15)
```
