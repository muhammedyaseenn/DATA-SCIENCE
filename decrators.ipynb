```python
#decrator
def div(a,b):
  print(a/b)

div(4,2)

div(2,4)
```

```python
def div(a,b):
  if a<b:
    a,b=b,a #swapping
  print(a/b)

div(2,4)
```

```python
#normal function
def div(a,b):
  print(a/b)

#adding decorators
def magic_div(fun):

  def inner(a,b): #inner() is the wrapper that adds new behavior
    if a<b:
      a,b=b,a
    return fun(a,b)
  return inner

div2= magic_div(div) #div=magic_div(div)
                     #div(2,4)
div2(2,4)
```

```python
#normal function
def greet(name):
  return f"Hello,{name}!"

#decorator to convert result to uppercase
def uppercase_decorator(func):
  def inner(name):
    result = func(name)
    return result.upper()
  return inner


#apply decorator manually
greet_upper = uppercase_decorator(greet)

#call the decorated function
print(greet_upper("Alice"))
```
